```python
%sql
CREATE CATALOG IF NOT EXISTS brasileirao
COMMENT 'Brasileirão Catalog'
```

```python
%sql
CREATE SCHEMA IF NOT EXISTS brasileirao.bronze
COMMENT 'Bronze schema for raw data'
```

```python
%sql
CREATE SCHEMA IF NOT EXISTS brasileirao.silver
COMMENT 'Silver schema for clean data'
```

```python
%sql
CREATE SCHEMA IF NOT EXISTS brasileirao.gold
COMMENT 'Gold schema for insights and ML'
```

```python
%sql
CREATE VOLUME brasileirao.bronze.raw_files
COMMENT 'Volume for initial raw ingestion files'
```

```python
%sql
CREATE OR REPLACE TABLE brasileirao.bronze.matches
COMMENT "Bronze table with raw data"
AS
SELECT *
FROM read_files(
    '/Volumes/brasileirao/bronze/raw_files/matches',
    format => 'csv'
)
```

```python
%sql
CREATE OR REPLACE TABLE brasileirao.bronze.matches_raw
COMMENT "Bronze table with raw data"
AS
SELECT *
FROM read_files(
    '/Volumes/brasileirao/bronze/raw_files/matches_raw',
    format => 'csv'
)
```
