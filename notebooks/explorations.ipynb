```python
%sql
USE CATALOG brasileirao;
USE SCHEMA gold
```

```python
%sql
SELECT *
FROM gold.matches_metrics
```

```python
%sql
DESCRIBE TABLE gold.matches_metrics
```

```python
%sql
SELECT
    COUNT(*) AS total_partidas,

    COUNT(DISTINCT partida_id) AS partidas_unicas,

    SUM(CASE WHEN partida_id IS NULL THEN 1 ELSE 0 END)
        AS ids_nulos,

    SUM(CASE WHEN mandante_id IS NULL OR visitante_id IS NULL
        THEN 1 ELSE 0 END) AS clubes_nulos,

    SUM(CASE WHEN gols_mandante IS NULL OR gols_visitante IS NULL
        THEN 1 ELSE 0 END) AS resultados_nulos,

    MIN(data) AS primeira_partida,
    MAX(data) AS ultima_partida

FROM gold.matches_metrics
```
