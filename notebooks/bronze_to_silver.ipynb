# Join to get the team names

```python
%sql
USE CATALOG ${catalog};
USE SCHEMA ${schema_silver};

CREATE OR REPLACE TABLE ${catalog}.${schema_silver}.matches_teams
USING DELTA
AS
SELECT
    m.partida_id,
    m.id_fonte,
    m.temporada,
    m.data,
    m.mandante_id,
    tm.nome AS nome_mandante,
    m.visitante_id,
    tv.nome AS nome_visitante,
    m.gols_mandante,
    m.gols_visitante

FROM ${catalog}.${schema_bronze}.matches m

LEFT JOIN ${catalog}.${schema_bronze}.teams tm
    ON m.mandante_id = tm.clube_id

LEFT JOIN ${catalog}.${schema_bronze}.teams tv
    ON m.visitante_id = tv.clube_id
```
