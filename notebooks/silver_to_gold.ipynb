```python
%sql
USE CATALOG ${catalog};
USE SCHEMA ${schema_gold};

CREATE OR REPLACE TABLE ${catalog}.${schema_gold}.matches_metrics
USING DELTA
AS
SELECT
    m.temporada,
    m.data,
    m.nome_mandante,
    m.nome_visitante,
    m.gols_mandante,
    m.gols_visitante,
    m.partida_id,
    m.mandante_id,
    m.visitante_id

FROM ${catalog}.${schema_silver}.matches_teams AS m

WHERE m.data >= DATE '2021-01-01'
  AND m.data < DATE_ADD(CURRENT_DATE(), 1);
```
